<h2 style="color:DodgerBlue">Название проекта:</h2>
Создать базовый класс Book в С#, который будет представлять информацию о книгах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

### Вариант задания 5


<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;

// Базовый класс Book
public class Book
{
    // Атрибуты
    public string Title { get; set; }
    public string Author { get; set; }
    public int YearOfPublication { get; set; }

    // Конструктор
    public Book(string title, string author, int yearOfPublication)
    {
        Title = title;
        Author = author;
        YearOfPublication = yearOfPublication;
    }

    // Метод для получения информации о книге
    public virtual string GetInfo()
    {
        return $"Книга: \"{Title}\", Автор: {Author}, Год издания: {YearOfPublication}";
    }

    // Метод для вывода сообщения о чтении книги
    public virtual void Read()
    {
        Console.WriteLine($"Чтение книги: \"{Title}\"");
    }

    // Метод для вывода сообщения о выдаче книги на чтение
    public virtual void Borrow()
    {
        Console.WriteLine($"Книга \"{Title}\" выдана на чтение");
    }
}

// Производный класс Учебник (Textbook)
public class Textbook : Book
{
    // Дополнительный атрибут
    public string Subject { get; set; }

    // Конструктор
    public Textbook(string title, string author, int yearOfPublication, string subject) 
        : base(title, author, yearOfPublication)
    {
        Subject = subject;
    }

    // Переопределение метода Read() с добавлением информации о предмете
    public override void Read()
    {
        Console.WriteLine($"Изучение учебника по предмету '{Subject}': \"{Title}\"");
    }

    // Переопределение метода GetInfo() для включения информации о предмете
    public override string GetInfo()
    {
        return base.GetInfo() + $", Предмет: {Subject}";
    }
}

// Производный класс Художественная литература (Fiction)
public class Fiction : Book
{
    // Дополнительный атрибут
    public string Genre { get; set; }

    // Конструктор
    public Fiction(string title, string author, int yearOfPublication, string genre) 
        : base(title, author, yearOfPublication)
    {
        Genre = genre;
    }

    // Переопределение метода Borrow() с добавлением информации о жанре
    public override void Borrow()
    {
        Console.WriteLine($"Художественная книга в жанре '{Genre}' \"{Title}\" выдана на чтение");
    }

    // Переопределение метода GetInfo() для включения информации о жанре
    public override string GetInfo()
    {
        return base.GetInfo() + $", Жанр: {Genre}";
    }
}

// Производный класс Научная литература (ScientificLiterature)
public class ScientificLiterature : Book
{
    // Дополнительный атрибут
    public string FieldOfScience { get; set; }

    // Конструктор
    public ScientificLiterature(string title, string author, int yearOfPublication, string fieldOfScience) 
        : base(title, author, yearOfPublication)
    {
        FieldOfScience = fieldOfScience;
    }

    // Переопределение метода GetInfo() для включения информации об области науки
    public override string GetInfo()
    {
        return base.GetInfo() + $", Область науки: {FieldOfScience}";
    }

    // Новый метод, специфичный для научной литературы
    public void Research()
    {
        Console.WriteLine($"Проведение исследования по книге \"{Title}\" в области {FieldOfScience}");
    }
}

// Демонстрационная программа
class Program
{
    static void Main(string[] args)
    {
        Console.WriteLine("=== Демонстрация наследования и полиморфизма ===\n");

        // Создание объектов разных типов
        Book book1 = new Book("Основы программирования", "Иванов И.И.", 2020);
        Textbook textbook1 = new Textbook("Математика", "Петров П.П.", 2021, "Математика");
        Fiction fiction1 = new Fiction("Преступление и наказание", "Достоевский Ф.М.", 1866, "Роман");
        ScientificLiterature science1 = new ScientificLiterature("Краткая история времени", "Хокинг С.", 1988, "Физика");

        // Массив книг для демонстрации полиморфизма
        Book[] books = { book1, textbook1, fiction1, science1 };

        // Демонстрация полиморфизма - вызов методов через базовый класс
        Console.WriteLine("=== Информация о всех книгах ===\n");
        foreach (var book in books)
        {
            Console.WriteLine(book.GetInfo());
        }

        Console.WriteLine("\n=== Демонстрация метода Read() ===\n");
        foreach (var book in books)
        {
            book.Read();
        }

        Console.WriteLine("\n=== Демонстрация метода Borrow() ===\n");
        foreach (var book in books)
        {
            book.Borrow();
        }

        // Демонстрация специфических методов производных классов
        Console.WriteLine("\n=== Специфические методы ===\n");
        
        // Для Textbook
        textbook1.Read(); // Переопределенный метод
        
        // Для Fiction
        fiction1.Borrow(); // Переопределенный метод
        
        // Для ScientificLiterature
        science1.Research(); // Новый метод

        // Демонстрация приведения типов
        Console.WriteLine("\n=== Демонстрация приведения типов ===\n");
        
        if (books[1] is Textbook textbook)
        {
            Console.WriteLine($"Это учебник по предмету: {textbook.Subject}");
        }

        if (books[2] is Fiction fiction)
        {
            Console.WriteLine($"Это художественная книга жанра: {fiction.Genre}");
        }
    }
}
```
